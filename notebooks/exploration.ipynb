# Dual-Engine Banking Chatbot — Exploratory Analysis

This notebook explores the Banking77 dataset and validates the training pipeline.


```python
import sys
sys.path.insert(0, '..')

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_config, load_json

cfg = load_config('../configs/config.yaml')
print('Config loaded.')
```

## 1. Dataset Overview

```python
train_rows = load_json('../data/train.json')
test_rows  = load_json('../data/test.json')
label_names = load_json('../data/label_names.json')

print(f'Train examples : {len(train_rows):,}')
print(f'Test  examples : {len(test_rows):,}')
print(f'Number of intents: {len(label_names)}')
print('\nSample intents:', label_names[:10])
```

## 2. Class Distribution

```python
train_df = pd.DataFrame(train_rows)
train_df['intent'] = train_df['label'].apply(lambda x: label_names[x])

counts = train_df['intent'].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
counts.plot(kind='hist', bins=20, ax=ax, color='steelblue', edgecolor='white')
ax.set_title('Distribution of examples per class (train)')
ax.set_xlabel('Number of examples')
ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()

print(f'Mean examples/class : {counts.mean():.1f}')
print(f'Min examples/class  : {counts.min()}')
print(f'Max examples/class  : {counts.max()}')
```

## 3. Sample Texts

```python
train_df.groupby('intent').head(2)[['intent', 'text']].head(20)
```

## 4. Text Length Analysis

```python
train_df['word_count'] = train_df['text'].apply(lambda x: len(x.split()))

fig, ax = plt.subplots(figsize=(8, 4))
train_df['word_count'].hist(bins=30, ax=ax, color='coral', edgecolor='white')
ax.set_title('Word count distribution (train)')
ax.set_xlabel('Words per query')
ax.set_ylabel('Count')
plt.tight_layout()
plt.show()

print(train_df['word_count'].describe())
```
